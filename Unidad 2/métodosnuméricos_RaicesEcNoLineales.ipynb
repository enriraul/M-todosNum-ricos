```python
import sympy as sp
```

```python
#
#  Esta función nos permite ajustar cualquier función no lineal en formato
#  texto y permitir tenerla como una expresión que pueda ser empleada
#  para evaluar valores en especifico. Esto nos permitirá usarla previa a cada
#  método que utilicemos y no tengamos que especificar parametro a parametro.
#  Cabe mencionar que esta función nos permite obtener tanto la expresión
#  matemática como su derivada que será empleada en algunos métodos
#

def crear_funcion(expresion):
    x = sp.Symbol('x')

    f_simbolica = sp.sympify(expresion)

    f = sp.lambdify(x, f_simbolica, modules="numpy")

    df_simbolica = sp.diff(f_simbolica, x)
    df = sp.lambdify(x, df_simbolica, modules="numpy")

    return f, df
```

```python
def biseccion(f, a, b, tol, max_iter):

    if f(a) * f(b) >= 0:
        raise ValueError("El intervalo no cumple el cambio de signo.")

    historial = []

    for i in range(1, max_iter + 1):

        xr = (a + b) / 2
        error = abs(b - a) / 2

        historial.append({
            "iteracion": i,
            "a": a,
            "b": b,
            "xr": xr,
            "f(xr)": f(xr),
            "error": error
        })

        if abs(f(xr)) < tol or error < tol:
            break

        if f(a) * f(xr) < 0:
            b = xr
        else:
            a = xr

    return xr, historial
```

```python
def newton_raphson(f, df, x0, tol, max_iter):

    historial = []

    for i in range(1, max_iter + 1):

        if abs(df(x0)) < 1e-14:
            raise ValueError("La derivada es cero o demasiado pequeña.")

        xr = x0 - f(x0) / df(x0)
        error = abs(xr - x0)

        historial.append({
            "iteracion": i,
            "x_anterior": x0,
            "xr": xr,
            "f(xr)": f(xr),
            "error": error
        })

        if abs(f(xr)) < tol or error < tol:
            break

        x0 = xr

    return xr, historial
```

```python
def secante(f, x0, x1, tol, max_iter):

    historial = []

    for i in range(1, max_iter + 1):

        denominador = f(x1) - f(x0)

        if abs(denominador) < 1e-14:
            raise ValueError("División entre cero o denominador muy pequeño.")

        xr = x1 - f(x1) * (x1 - x0) / denominador

        error = abs(xr - x1)

        historial.append({
            "iteracion": i,
            "x0": x0,
            "x1": x1,
            "xr": xr,
            "f(xr)": f(xr),
            "error": error
        })

        if abs(f(xr)) < tol or error < tol:
            break

        x0, x1 = x1, xr

    return xr, historial
```

```python
def punto_fijo(g, x0, tol, max_iter):

    historial = []

    for i in range(1, max_iter + 1):

        xr = g(x0)
        error = abs(xr - x0)

        historial.append({
            "iteracion": i,
            "x_anterior": x0,
            "xr": xr,
            "error": error
        })

        if error < tol:
            break

        x0 = xr

    return xr, historial
```
